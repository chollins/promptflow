```python
from pathlib import Path

from form_loader import load_prompt_form
from prompt_renderer import render_prompt
from providers.openai_provider import execute_prompt
```

```python

def collect_inputs(form):

    values = {}

    print("\nEnter values:\n")

    for field in form.fields:

        value = input(
            f"{field.label}: "
        )

        values[field.id] = value

    return values
```

```python
PROJECT_ROOT = Path(__file__).resolve().parent.parent

FORM_PATH = PROJECT_ROOT / "forms" / "customer-summary.form.json"

form = load_prompt_form(FORM_PATH)
user_values = collect_inputs(form)


rendered_prompt = render_prompt(
    form.prompt.user,
    user_values
)


print("\nGenerated Prompt")
print("----------------")
print(rendered_prompt)


print("\nCalling LLM...\n")


result = execute_prompt(
    form.prompt.system,
    rendered_prompt
)


print("==============================")
print("LLM RESPONSE")
print("==============================")

print(result)
```
